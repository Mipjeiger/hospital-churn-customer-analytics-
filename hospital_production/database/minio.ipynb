```python
import os
import boto3
from dotenv import load_dotenv

# Load your local .env configuration
load_dotenv()

# MinIO endpoint and credentials setup
ENDPOINT_URL = os.getenv("MLFLOW_S3_ENDPOINT_URL", "http://localhost:9000")
ACCESS_KEY = os.getenv("AWS_ACCESS_KEY_ID")
SECRET_KEY = os.getenv("AWS_SECRET_ACCESS_KEY")
BUCKET_NAME = "hospital-models"  # Change this to your specific bucket name if different

# Initialize the S3 client pointing to your local MinIO container
s3_client = boto3.client(
    "s3",
    endpoint_url=ENDPOINT_URL,
    aws_access_key_id=ACCESS_KEY,
    aws_secret_access_key=SECRET_KEY,
    config=boto3.session.Config(signature_version="s3v4")
)

def list_minio_models(bucket_name, prefix=""):
    print(f"🔍 Scanning bucket '{bucket_name}' for models/artifacts...\n")
    try:
        paginator = s3_client.get_paginator('list_objects_v2')
        page_iterator = paginator.paginate(Bucket=bucket_name, Prefix=prefix)
        
        count = 0
        for page in page_iterator:
            if "Contents" in page:
                for obj in page["Contents"]:
                    print(f"📂 Path: {obj['Key']} | Size: {obj['Size']:,} bytes")
                    count += 1
            else:
                print("⚠️ No objects found under this path.")
                
        print(f"\n✨ Total items found: {count}")
        
    except Exception as e:
        print(f"❌ Error connecting to MinIO: {e}")

# Run the function
if __name__ == "__main__":
    list_minio_models(BUCKET_NAME)
```
